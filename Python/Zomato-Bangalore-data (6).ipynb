## 📌 Introduction

- The Zomato Restaurant Dataset contains detailed information about restaurants and food orders across major cities in India, with a strong focus on Bangalore.
- The dataset includes attributes related to restaurant details, location, cuisine, pricing, customer ratings, and order dates, making it suitable for analyzing customer behavior and restaurant performance in the food delivery domain.

- This dataset helps in understanding how factors such as pricing, location, online ordering availability, table booking options, and customer engagement (ratings and reviews) influence restaurant success on food delivery platforms.

## 📊 Dataset Description

- The dataset consists of multiple columns, including:
- Restaurant Name – Name of the restaurant
- Location / City / State – Geographic details of the restaurant
- Category / Cuisine – Type of food offered
- Dish Name – Individual food items
- Price (INR) – Cost of the dish
- Rating – Customer rating of the restaurant or dish
- Rating Count / Votes – Number of customer reviews
- Order Date – Date when the order was placed

## 🎯 Objective of the Analysis

### The main objectives of this project are:

- To clean and preprocess real-world restaurant data with missing values and inconsistent formats
- To analyze customer preferences and restaurant performance
- To identify key factors influencing ratings and pricing
- To extract business-driven insights that can help food delivery platforms and restaurant owners make data-informed decisions

###  Drop Irrelevant Columns

```python
df.drop(columns =['url','address','phone','dish_liked','menu_item','reviews_list'],inplace = True)
```

```python
df.head()
```

### Rename Columns for Consistency

```python
df.columns
```

```python
df.rename(columns ={'approx_cost(for two people)':'cost' ,
                   'listed_in(type)':'type ',
                    'listed_in(city)':'city'} , inplace = True)
```

```python
df.head()
```

## Handling missing Values

```python
df.isnull().sum()
```

#### Clean Rate Column

```python
df['rate'] = df['rate'].replace({'NEW':np.nan ,'-':np.nan})

df.dropna(how = 'any' , inplace = True)
```

```python
df.shape
```

```python
df.isnull().sum()
```

### Convert Data Types

```python
df['cost'].sample(10)
```

```python
df['rate'] = df['rate'].apply(lambda x : str(x).split('/')[0])
df['rate'] = df['rate'].astype(float)

df['cost'] = df['cost'].apply(lambda x : str(x).replace(',',''))
df['cost'] = df['cost'].astype(float)
```

```python
df['rate'].sample(5)
```

## Remove Duplicates

```python
df.duplicated().sum()
```

```python
# Romove duplicates Rows
df.drop_duplicates(inplace =True)
```

```python
df.info()
```

```python
df.output_csv("Zomato_Clean_Data.csv")
```

## Exploratory Data Analysis

### Distribution of Ratings

```python
plt.figure(figsize = (6,4))
sns.histplot(df['rate'] , bins =20 ,kde = True , color = 'blue')
plt.title("Distribution of Restaurant Ratings")  
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.show()
```

### Insight
- Most restaurants have ratings between 3.5 to 4.0
- Very few restaurants have ratings:
- below 3.0 (low rated)
- above 4.5 (excellent)

### Online Orders vs Ratings

```python
plt.figure(figsize =(8,6))
sns.boxplot(x ='online_order' , y ='rate' , data=df)
plt.title('Online Orders vs Ratings')
plt.xlabel('Online Orders')
plt.ylabel('Ratings')
plt.show()
```

### Insight
- Restaurants that allow online orders mostly have ratings around 3.6 – 3.9
- Restaurants that do not allow online orders also have ratings around 3.4 – 3.8
- But online-order restaurants show slightly better ratings overall.

### Table Booking vs Ratings

```python
plt.figure(figsize =(8,6))
sns.boxplot(x ='book_table' , y ='rate' , data=df)
plt.title('Table Booking vs Ratings')
plt.xlabel('Table Booking')
plt.ylabel('Ratings')
plt.show()
```

### Insight:
- Restaurants that have table booking = Yes generally have higher ratings.
- Restaurants with table booking = No have slightly lower ratings and more low-rated restaurants.

## Top 10 Restaurant Locations

```python
# Get top 25 comman restaurant types
top25 = df['rest_type'].value_counts().nlargest(25).index

# Filter dataset for only these top 25
df_top = df[df['rest_type'].isin(top25)]

plt.figure(figsize = (14,10))
sns.countplot(y ='rest_type', data =df_top , order =top25, palette = 'coolwarm')
plt.title("Top 25 Most Comman Restaurant Types" , fontsize= 14 , fontweight = 'bold')
plt.xlabel("Count")
plt.ylabel("Restaurant Types")
plt.show()
```

### Insight
- The most common restaurant types are:
- Quick Bites (highest)
- Casual Dining (second highest)
- Other types like café, desserts, fine dining etc. are less compared to these.

### Restaurant Types Distribution

```python
# Get top 10 locations with the most restaurants
top10 = df['location'].value_counts().nlargest(10).index

# Filter dataset for only these top 25
df_top = df[df['location'].isin(top10)]

plt.figure(figsize = (12,6))
sns.countplot(x ='location', data =df_top , order =top10, palette = 'viridis')
plt.title("Top 10 locations with the most restaurants" , fontsize= 14 , fontweight = 'bold')
plt.xlabel("Location")
plt.ylabel("Numbers of Resturants")
plt.xticks(rotation = 45)
plt.show()
```

### Insights
- Some locations have many restaurants and high demand.
- These areas are food hotspots.

## Restaurant Type vs Average Votes

```python
rest_type_votes = (df.groupby("rest_type")["votes"].mean().sort_values(ascending=False).head(10))

plt.figure(figsize=(10,5))
rest_type_votes.plot(kind="bar")
plt.title("Restaurant Type vs Average Votes")
plt.xlabel("Restaurant Type")
plt.ylabel("Average Votes")
plt.xticks(rotation=90)
plt.show()
```

### Insights
- Casual Dining and Quick Bites receive more customer votes.
-  Customers prefer simple and comfortable dining.

### Low Rating BUt High Votes

### Insights
- Some restaurants are popular but they have low ratings.
- These restaurants need quality improvement.
